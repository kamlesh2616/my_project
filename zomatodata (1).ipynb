```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("Zomato data .csv")
df
```

```python
df.info()
```

```python
df.isna().sum()
```

```python
# convert the data type of column - rate
```

```python

def handlerate(value):
    value = str(value).split('/')
    value = value[0];
    return float(value)

df["rate"] = df["rate"].apply(handlerate)
df
```

```python
df.info()
```

```python
# type of resturant
```

```python
sns.countplot(x=df["listed_in(type)"])
plt.xlabel("type of resturant")
```

```python
# conclusion - majority of the rasturant falls in dinning category
```

```python
df.head()
```

```python
grouped_data = df.groupby("listed_in(type)")["votes"].sum()
result = pd.DataFrame({"votes" : grouped_data})
plt.plot(result,c="green", marker="o")
plt.xlabel("type of resturant", c="red", size=20)
plt.ylabel("votes", c="red", size=20)
```

```python
df.head()
```

```python
linechart = df[["listed_in(type)","votes"]].groupby("listed_in(type)")[["votes"]].sum()
linechart
```

```python
sns.lineplot(x="listed_in(type)",y="votes",data=linechart,color="green",marker="o")
```

```python
plt.hist(df["rate"],bins=2)
plt.title("ratings distribution")
plt.show()
```
